```python
import pandas as pd
import spacy

# 1. Load the spaCy language brain
nlp = spacy.load("en_core_web_sm")

# This is our target online URL
dataset_url = "https://raw.githubusercontent.com/fenago/datasets/main/twitter_training.csv"

print("Attempting to load data from the internet...")

try:
    # TRY BLOCK: Attempt to read from the web link
    # We use nrows=50 to only fetch the first 50 lines to keep it fast
    df_online = pd.read_csv(dataset_url, names=['id', 'entity', 'sentiment', 'text'], nrows=50)
    headlines = df_online['text'].dropna().head(15).tolist()
    print("Success! Live online dataset loaded successfully.")

except Exception as error:
    # EXCEPT BLOCK: If the link is dead, this code runs automatically instead of crashing
    print(f"\n[Notice] Web link unavailable (Error: {error}).")
    print("Switching to local backup text dataset automatically...")

    # Backup dataset containing rich entity examples
    headlines = [
        "Tesla Inc. CEO Elon Musk announced a new Gigafactory construction in Berlin, Germany.",
        "Apple scheduled its keynote event for next Monday in Cupertino, California.",
        "The Central Bank of Nigeria updated its financial guidelines in Abuja this January.",
        "Nvidia corp reported record-breaking profits due to surging AI chip demands in 2026.",
        "Aliko Dangote met with trade ministers in Nairobi, Kenya last week."
    ]

# 2. Extract Named Entities
live_extracted_data = []

for text in headlines:
    doc = nlp(text)
    for ent in doc.ents:
        live_extracted_data.append({
            "Source Text": text[:45] + "...",
            "Entity Found": ent.text,
            "Type Label": ent.label_,
            # FIX: Changed 'explanation' to 'explain'
            "Description": spacy.explain(ent.label_)
        })

# 3. Output results into a structured DataFrame
df_results = pd.DataFrame(live_extracted_data)

print("\n--- NER EXTRACTION RESULTS ---")
print(df_results)

# 3. Output results into a structured DataFrame
df_results = pd.DataFrame(live_extracted_data)

print("\n--- NER EXTRACTION RESULTS ---")
print(df_results)
```

```python
import sys
import subprocess


try:
    import gtts
except ModuleNotFoundError:
    print("Installing gTTS package directly inside the runtime environment...")
    # This runs the pip command inside your active session safely
    subprocess.check_call([sys.executable, "-m", "pip", "install", "gTTS"])
    import gtts

from gtts import gTTS
from IPython.display import Audio, display


sample_text = "Welcome to the dsn nlp week 8 assignment 2. Artificial intelligence transforms how we process human communication."

print("\n--- Tool 1: Generating Audio using Google TTS (gTTS) ---")
try:
    # Initialize the neural cloud engine
    gtts_engine = gTTS(text=sample_text, lang='en', tld='com', slow=False)
    gtts_filename = "google_neural_output.mp3"
    gtts_engine.save(gtts_filename)

    print("Google TTS Audio saved successfully! Click play below:")
    display(Audio(gtts_filename))
except Exception as e:
    print(f"Cloud TTS execution encountered an error: {e}")
```
